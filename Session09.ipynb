    <div style="font-size:16.5px;">از کاربر سن، نوع بلیت و وضعیت عضویت را بگیر. قانون‌ها:<br>۱) اگر بلیت <code dir='ltr'>vip</code> باشد و کاربر banned نباشد → VIP Gate.<br>۲) اگر سن حداقل 12 باشد و بلیت regular یا vip باشد → Main Gate.<br>۳) اگر سن کمتر از 12 ولی عضویت family باشد → Family Gate.<br>۴) در غیر این صورت ورود رد شود.</div>

```python
age = int(input("Age: "))
ticket = input("Ticket (regular/vip/none): ").strip().lower()
membership = input("Membership (family/normal): ").strip().lower()
is_banned = input("Banned? yes/no: ").strip().lower() == "yes"

if ticket == "vip" and not is_banned:
    print("⭐ VIP Gate")
elif age >= 12 and ticket in ("regular", "vip") and not is_banned:
    print("✅ Main Gate")
elif age < 12 and membership == "family" and not is_banned:
    print("👨‍👩‍👧 Family Gate")
else:
    print("⛔ Entry denied")
```
