    <div style="font-size:16.5px;"><code dir='ltr'>remove(value)</code> بر اساس مقدار حذف می‌کند. <code dir='ltr'>pop(index)</code> عضو را حذف و برمی‌گرداند. <code dir='ltr'>del</code> با index یا slice حذف می‌کند.</div>

```python
items = ["a", "b", "c", "d"]

items.remove("b")
last_item = items.pop()
del items[0]

print(items)
print("popped:", last_item)
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'Vazirmatn','Noto Sans Arabic','Tahoma','Segoe UI',Arial,sans-serif;line-height:2;">
  <div style="border:1px solid #d9e0eb;border-right:5px solid #36B27B;border-radius:14px;padding:16px 18px;margin:10px 0 18px 0;">
    <div style="display:flex;gap:10px;align-items:center;justify-content:flex-start;flex-wrap:wrap;margin-bottom:8px;">
      <strong style="font-size:21px;">حرکت روی لیست — Traversal</strong><span style="background:#36B27B18;color:#36B27B;padding:4px 10px;border-radius:999px;font-weight:700;">Traversal</span>
    </div>
    <div style="font-size:16.5px;">رایج‌ترین راه حرکت روی لیست، <code dir='ltr'>for item in list</code> است. اگر index هم لازم باشد از <code dir='ltr'>enumerate()</code> استفاده کن.</div>
  </div>
  <hr style="border:0;border-top:1px solid rgba(120,130,150,.28);margin:20px 0;">
</div>

```python
players = ["Sara", "Ali", "Nima"]

for player in players:
    print(player)

print("---")

for index, player in enumerate(players):
    print(index, player)
```

```python
inventory = ["sword", "potion"]

while True:
    command = input("Command (show/add/remove/count/quit): ").strip().lower()

    if command == "show":
        for index, item in enumerate(inventory, start=1):
            print(index, item)

    elif command == "add":
        item = input("Item to add: ").strip().lower()
        inventory.append(item)
        print("Added:", item)

    elif command == "remove":
        item = input("Item to remove: ").strip().lower()
        if item in inventory:
            inventory.remove(item)
            print("Removed:", item)
        else:
            print("Item not found")

    elif command == "count":
        print("Items:", len(inventory))

    elif command == "quit":
        print("Inventory saved in memory for this run. Bye!")
        break

    else:
        print("Unknown command")
```
