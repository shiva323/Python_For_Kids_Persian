```python
pizza = 7
people = 2
print("تقسیم دقیق:", pizza / people)
print("تکه کامل:", pizza // people)
print("باقی‌مانده:", pizza % people)
```

```python
import math
radius = float(input("شعاع دایره: "))
area = math.pi * radius ** 2
circumference = 2 * math.pi * radius
print(f"مساحت: {round(area, 2)}")
print(f"محیط: {round(circumference, 2)}")
```
