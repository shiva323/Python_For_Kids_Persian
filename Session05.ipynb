    <div style="font-size:16.5px;">چون <code dir='ltr'>input()</code> متن برمی‌گرداند، ورودی را با <code dir='ltr'>float()</code> به عدد تبدیل می‌کنیم تا هم عدد صحیح و هم اعشاری قبول شود.</div>

```python
a = float(input("عدد اول: "))
b = float(input("عدد دوم: "))

print(f"a = {a}")
print(f"b = {b}")
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'Vazirmatn','Noto Sans Arabic','Tahoma','Segoe UI',Arial,sans-serif;line-height:2;">
  <div style="border:1px solid #d9e0eb;border-right:5px solid #815BDA;border-radius:14px;padding:16px 18px;margin:10px 0 18px 0;">

      <strong style="font-size:21px;">قبل از اجرا حدس بزن</strong><span style="background:#815BDA18;color:#815BDA;padding:4px 10px;border-radius:999px;font-weight:700;">پیش‌بینی</span>
    </div>
    <div style="font-size:16.5px;">اگر کاربر برای عدد اول <code dir='ltr'>12.5</code> و برای عدد دوم <code dir='ltr'>4</code> وارد کند، نوع هر دو متغیر چیست؟</div>
  </div>
  <hr style="border:0;border-top:1px solid rgba(120,130,150,.28);margin:20px 0;">

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'Vazirmatn','Noto Sans Arabic','Tahoma','Segoe UI',Arial,sans-serif;line-height:2;">
  <div style="border:1px solid #d9e0eb;border-right:5px solid #F1973D;border-radius:14px;padding:16px 18px;margin:10px 0 18px 0;">
    <div style="display:flex;gap:10px;align-items:center;justify-content:flex-start;flex-wrap:wrap;margin-bottom:8px;">
      <strong style="font-size:21px;">قدم ۲: محاسبه همه عملیات</strong><span style="background:#F1973D18;color:#F1973D;padding:4px 10px;border-radius:999px;font-weight:700;">منطق</span>
    </div>
    <div style="font-size:16.5px;">یک بار محاسبه می‌کنیم و نتیجه‌ها را داخل متغیرهای معنی‌دار نگه می‌داریم. این کار کد را برای خواننده واضح‌تر می‌کند.</div>
  </div>
  <hr style="border:0;border-top:1px solid rgba(120,130,150,.28);margin:20px 0;">
</div>

```python
sum_result = a + b
diff_result = a - b
product_result = a * b
division_result = a / b
floor_div_result = a // b
remainder_result = a % b
power_result = a ** b
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'Vazirmatn','Noto Sans Arabic','Tahoma','Segoe UI',Arial,sans-serif;line-height:2;">
  <div style="border:1px solid #d9e0eb;border-right:5px solid #2FA86E;border-radius:14px;padding:16px 18px;margin:10px 0 18px 0;">
    <div style="display:flex;gap:10px;align-items:center;justify-content:flex-start;flex-wrap:wrap;margin-bottom:8px;">
      <strong style="font-size:21px;">قدم ۳: خروجی خوانا</strong><span style="background:#2FA86E18;color:#2FA86E;padding:4px 10px;border-radius:999px;font-weight:700;">نمایش</span>
    </div>
    <div style="font-size:16.5px;">نتیجه‌ها را با f-string چاپ می‌کنیم و برای بعضی خروجی‌ها از <code dir='ltr'>round(..., 2)</code> استفاده می‌کنیم.</div>
  </div>
  <hr style="border:0;border-top:1px solid rgba(120,130,150,.28);margin:20px 0;">
</div>

```python
print("🧮 نتیجه‌ها")
print(f"{a} + {b} = {sum_result}")
print(f"{a} - {b} = {diff_result}")
print(f"{a} × {b} = {product_result}")
print(f"{a} ÷ {b} = {round(division_result, 2)}")
print(f"{a} // {b} = {floor_div_result}")
print(f"{a} % {b} = {remainder_result}")
print(f"{a} ** {b} = {round(power_result, 2)}")
```

      <strong style="font-size:21px;">ارتقای ماشین‌حساب</strong><span style="background:#2FA86E18;color:#2FA86E;padding:4px 10px;border-radius:999px;font-weight:700;">چالش</span>
    </div>
    <div style="font-size:16.5px;">دو خروجی دیگر اضافه کن: بزرگ‌ترین عدد با <code dir='ltr'>max(a, b)</code> و فاصله دو عدد با <code dir='ltr'>abs(a - b)</code>.</div>
  </div>

```python
print(f"بزرگ‌تر: {max(a, b)}")
print(f"فاصله دو عدد: {abs(a - b)}")
```
