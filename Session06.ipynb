    <div style="font-size:16.5px;">اندیس‌ها از صفر شروع می‌شوند. در <code dir='ltr'>Python</code>، کاراکتر شماره صفر <code dir='ltr'>P</code> است. اندیس منفی از آخر رشته می‌شمارد.</div>

```python
text = "Python"
print(text[0])
print(text[1])
print(text[-1])
print(text[-2])
```

    <div style="display:flex;gap:10px;align-items:center;justify-content:flex-start;flex-wrap:wrap;margin-bottom:8px;">
      <strong style="font-size:21px;">Slicing: بریدن یک بخش</strong><span style="background:#815BDA18;color:#815BDA;padding:4px 10px;border-radius:999px;font-weight:700;">Slicing</span>
    </div>
    <div style="font-size:16.5px;">ساختار اصلی: <code dir='ltr'>text[start:stop]</code>. نقطه شروع حساب می‌شود ولی stop داخل نتیجه نیست.</div>
  </div>
  <hr style="border:0;border-top:1px solid rgba(120,130,150,.28);margin:20px 0;">
</div>

```python
text = "Computer"
print(text[0:3])
print(text[3:])
print(text[:4])
print(text[-3:])
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'Vazirmatn','Noto Sans Arabic','Tahoma','Segoe UI',Arial,sans-serif;line-height:2;">

    <div style="display:flex;gap:10px;align-items:center;justify-content:flex-start;flex-wrap:wrap;margin-bottom:8px;">
      <strong style="font-size:21px;">Step در slicing</strong><span style="background:#4A77FF18;color:#4A77FF;padding:4px 10px;border-radius:999px;font-weight:700;">Step</span>
    </div>
    <div style="font-size:16.5px;">فرم کامل‌تر <code dir='ltr'>text[start:stop:step]</code> است. step=2 یعنی یکی در میان؛ step=-1 یک ترفند معروف برای برعکس کردن رشته است.</div>
  </div>

```python
text = "abcdefgh"
print(text[::2])
print(text[::-1])
```
