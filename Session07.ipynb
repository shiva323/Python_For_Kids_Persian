```python
text = "hello python world"
print(text.upper())
print(text.lower())
print(text.title())
print(text.capitalize())
```

  <div style="border:1px solid #d9e0eb;border-right:5px solid #4A77FF;border-radius:14px;padding:16px 18px;margin:10px 0 18px 0;">

      <strong style="font-size:21px;">پروژه: Username Cleaner</strong><span style="background:#4A77FF18;color:#4A77FF;padding:4px 10px;border-radius:999px;font-weight:700;">پروژه</span>
    </div>
    <div style="font-size:16.5px;">کاربر نامی با فاصله و حروف نامرتب وارد می‌کند. برنامه آن را تمیز می‌کند: فاصله‌های ابتدا و انتها حذف، حروف کوچک و فاصله‌های داخلی به underscore تبدیل می‌شوند.</div>
  </div>

</div>

```python
raw_name = input("نام دلخواه را وارد کن: ")
username = raw_name.strip().lower().replace(" ", "_")

print(f"Username: {username}")
print(f"Length: {len(username)}")
```

      <strong style="font-size:21px;">ارتقای پروژه</strong><span style="background:#2FA86E18;color:#2FA86E;padding:4px 10px;border-radius:999px;font-weight:700;">چالش</span>
    </div>
    <div style="font-size:16.5px;">این موارد را هم چاپ کن: تعداد حرف <code dir='ltr'>a</code>، محل اولین underscore و اینکه username با underscore تمام می‌شود یا نه.</div>
  </div>
  <hr style="border:0;border-top:1px solid rgba(120,130,150,.28);margin:20px 0;">
</div>

```python
print("a count:", username.count("a"))
print("first _ index:", username.find("_"))
print("ends with _ :", username.endswith("_"))
```
