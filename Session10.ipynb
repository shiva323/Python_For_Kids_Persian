    <div style="font-size:16.5px;">اگر هم accumulator و هم counter داشته باشیم، می‌توانیم میانگین را بعد از حلقه محاسبه کنیم.</div>

      <strong style="font-size:21px;">Running Minimum و Maximum</strong><span style="background:#815BDA18;color:#815BDA;padding:4px 10px;border-radius:999px;font-weight:700;">Min / Max</span>
    </div>
    <div style="font-size:16.5px;">گاهی نمی‌خواهیم همه مقدارها را نگه داریم؛ فقط بهترین مقدار فعلی را نگه می‌داریم. اولین مقدار می‌تواند نقطه شروع min و max باشد.</div>
  </div>
  <hr style="border:0;border-top:1px solid rgba(120,130,150,.28);margin:20px 0;">

```python
current_min = None
current_max = None

for turn in range(5):
    value = float(input(f"Number {turn + 1}: "))

    if current_min is None or value < current_min:
        current_min = value

    if current_max is None or value > current_max:
        current_max = value

print("min:", current_min)
print("max:", current_max)
```

    </div>
    <div style="font-size:16.5px;">اگر هنگام حرکت روی یک iterable هم index و هم مقدار را بخواهیم، <code dir='ltr'>enumerate()</code> بسیار خواناست.</div>

```python
word = "CODE"

for index, char in enumerate(word):
    print(index, char)
```

    <div style="font-size:16.5px;">وقتی چیزی را پیدا کردیم و ادامه حلقه دیگر لازم نیست، <code dir='ltr'>break</code> حلقه را متوقف می‌کند.</div>

```python
secret = "python"

for attempt in range(1, 6):
    guess = input("Guess: ").strip().lower()
    if guess == secret:
        print("Found on attempt", attempt)
        break
```

    <div style="font-size:16.5px;"><code dir='ltr'>continue</code> بقیه کد همان iteration را رد می‌کند و مستقیم سراغ دور بعد می‌رود.</div>

```python
for number in range(1, 11):
    if number % 2 == 0:
        continue
    print(number)
```

    <div style="display:flex;gap:10px;align-items:center;justify-content:flex-start;flex-wrap:wrap;margin-bottom:8px;">
      <strong style="font-size:21px;">for یا while؟</strong><span style="background:#E68A3A18;color:#E68A3A;padding:4px 10px;border-radius:999px;font-weight:700;">انتخاب ابزار</span>
    </div>
    <div style="font-size:16.5px;">وقتی تعداد تکرار یا iterable مشخص است، معمولاً <code dir='ltr'>for</code> انتخاب طبیعی‌تری است. وقتی تکرار تا زمان برقرار بودن یک شرط ادامه دارد، <code dir='ltr'>while</code> مناسب‌تر است.</div>

```python
energy = 3

while energy > 0:
    print("Running... energy =", energy)
    energy -= 1

print("Rest!")
```

  <div style="border:1px solid #d9e0eb;border-right:5px solid #4A77FF;border-radius:14px;padding:16px 18px;margin:10px 0 18px 0;">
    <div style="display:flex;gap:10px;align-items:center;justify-content:flex-start;flex-wrap:wrap;margin-bottom:8px;">
      <strong style="font-size:21px;">پروژه: Score Analyzer</strong><span style="background:#4A77FF18;color:#4A77FF;padding:4px 10px;border-radius:999px;font-weight:700;">پروژه</span>
    </div>
    <div style="font-size:16.5px;">از کاربر بپرس چند نمره می‌خواهد وارد کند. سپس در یک حلقه: مجموع، تعداد قبولی‌های 50 به بالا، کمینه و بیشینه را نگه دار. در پایان میانگین را هم حساب کن. <b>بدون لیست</b>.</div>
  </div>
  <hr style="border:0;border-top:1px solid rgba(120,130,150,.28);margin:20px 0;">
</div>

```python
score_count = int(input("How many scores? "))

total = 0
passed = 0
lowest = None
highest = None

for i in range(score_count):
    score = float(input(f"Score {i + 1}: "))
    total += score

    if score >= 50:
        passed += 1

    if lowest is None or score < lowest:
        lowest = score

    if highest is None or score > highest:
        highest = score

average = total / score_count

print(f"Average: {average:.2f}")
print(f"Passed: {passed}")
print(f"Lowest: {lowest}")
print(f"Highest: {highest}")
```
